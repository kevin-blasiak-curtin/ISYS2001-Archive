## Why bother reflecting?

It is easy to treat the end of semester as the end of the learning. Research on how people learn is fairly consistent that the opposite is true: the act of looking back, putting into words what you did and why, is one of the most effective ways to make learning stick. Trying to recall something strengthens it far more than rereading it, and explaining your own thinking shows you where it is solid and where it is thin.

It is also a professional habit. Project teams in business and IT routinely run retrospectives and post-implementation reviews, because the lessons from a project are only useful if someone stops to write them down. Think of this notebook as a retrospective on your own learning.
