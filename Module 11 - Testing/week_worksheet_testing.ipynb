# Part 1: Reading and Writing Assertions (about 15 minutes)

An **assertion** is a line of code that states something that must be true. If it is true, Python says nothing and carries on. If it is false, Python stops with an `AssertionError`.

```python
assert some_condition, "message to show if it fails"
```

The message after the comma is optional but valuable. When a test fails, the message tells you what was supposed to happen.

One thing to keep in mind: **an assertion that fails stops the cell.** Any assertions below it do not run. This is why it helps to give each assertion a clear message, and why you fix failures one at a time.

The reverse is also true: a cell with no assertions in it runs without complaint, so "All assertions passed" only means something if your assertions are really there.

# Part 4: When a Bug Appears (about 20 minutes)

When a bug turns up, the instinct is to go straight to the code. A more reliable habit is to go to the specification first:

1. **Write a failing test** that captures the bug
2. **Describe the problem to AI clearly**, including the test
3. **Verify the fix** passes your test
4. **Keep the test**, so the bug cannot come back unnoticed (this is called a *regression test*)

## Exercise 4.1: Fixing a bug test-first

````python
def average_positive(numbers):
    """Calculate the average of the positive numbers only."""
    total = 0
    count = 0
    for num in numbers:
        if num > 0:
            total += num
        count += 1
    return total / count
````

## Step 1: Build a test plan

Real projects cannot be tested in one way. A good plan matches the **method** to the **thing being tested**.

| Method | Use it for | Example |
|---|---|---|
| **A: Assertions** | Logic with a predictable right answer | Calculations, validation, data cleaning, class methods that compute something |
| **M: Manual check** | Things you judge by looking | A Gradio interface, how a chart appears, whether messages are clear |
| **U: User check** | Whether it works for a real person | Asking a classmate to use your tool without your help |
| **N: Not testing** | Low risk, or not worth the effort | You can choose this, but you must say why |

Three situations need particular care in this unit:

- **Live API calls (Alpha Vantage, Gemini).** Do not put these inside assertions. The free tier has a small daily request limit, and the answers change from one run to the next. Test the code *around* the call instead: does your function handle a missing field, an empty response, or a bad ticker?
- **AI-generated text.** You cannot assert exact wording, because it varies. Assert properties instead: is a response returned, is it a string, is it not empty, does it have the structure your code expects?
- **Functions that do too much.** A function that fetches data, calculates something and prints the result is hard to assert on. Splitting out the calculation is often the first useful thing testing teaches you.
