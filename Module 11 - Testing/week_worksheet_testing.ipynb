# Part 1: The Idea in One Small Example

## Start with the requirement

A **requirement** says what something must do, in plain language. Here is one for an online shop:

> Orders of $100 or more ship free. Orders under $100 cost $9.95 to ship.

Here is a function that tries to meet it:

````python
def shipping_cost(order_total):
    """Return the shipping cost for an order."""
    if order_total >= 100:
        return 0.0
    return 9.95
````

## Turn the requirement into examples

An **assertion** is a line of code that checks one concrete example of the requirement. If the example is true, Python says nothing. If it is false, Python stops and shows the message.

```python
assert actual_result == expected_result, "message that explains what should happen"
```

Notice what we chose as examples: a **typical** case on each side of the rule, and the **boundary** where the rule changes (exactly $100, and just under it). Boundaries are where mistakes usually hide.

````python
assert shipping_cost(50) == 9.95,    "A $50 order is under $100, so shipping costs $9.95"
assert shipping_cost(150) == 0.0,    "A $150 order is over $100, so shipping is free"
assert shipping_cost(99.99) == 9.95, "Just under $100 still pays shipping"
assert shipping_cost(100) == 0.0,    "Exactly $100 ships free"

print("All assertions passed")
````

Everything passes, so the function matches the requirement for those four examples. Notice that a cell with no assertions would also print "All assertions passed", so the message only means something if the assertions are really there.

## What a failing test looks like

Now suppose someone "tidies up" the function and changes `>=` to `>`:

````python
def shipping_cost_buggy(order_total):
    """A version with a small mistake."""
    if order_total > 100:
        return 0.0
    return 9.95

# The typical examples still pass, so the bug is invisible to them:
assert shipping_cost_buggy(50) == 9.95
assert shipping_cost_buggy(150) == 0.0
print("Typical examples pass")

# Only the boundary example catches it. (try/except here only keeps the notebook running
# so you can see the message. In your own tests you do not need it.)
try:
    assert shipping_cost_buggy(100) == 0.0, "Exactly $100 ships free"
except AssertionError as error:
    print("FAILED:", error)
````

## Three things to take from this

1. **The requirement is the authority.** The assertions are examples that check the code against it. If an assertion and the requirement disagree, go back to the requirement.
2. **Choose examples deliberately.** A typical case, a boundary, and something awkward will tell you more than ten similar typical cases.
3. **Requirements have gaps.** What should happen if `order_total` is -20? The requirement does not say. Someone has to decide, and writing tests is often when you notice.
